Q3. Regular Expressions and Text Spliƫng
1. Take their original text from QuesƟon 1.

3. Use text spliƫng techniques to:
a. Split the text into words containing only alphabets (removing digits and special
characters).
b. Extract words starƟng with a vowel.

```python
# Regex Guide

# Component	Meaning	Example
# .	Any character except newline	a.b matches acb, arb, etc.
# \w	Word character [a-zA-Z0-9_]	\w+ matches a whole word
# \W	Non-word character	Used in re.split(r'\W+', text)
# \d	Digit [0-9]	\d+ matches numbers
# \D	Non-digit
# \s	Whitespace (space, tab, newline)
# \S	Non-whitespace
# \b	Word boundary	\bcat\b matches cat but not catalog
# ^	Start of string	^Hello matches strings that start with "Hello"
# $	End of string	world$ matches strings that end with "world"
# [abc]	One of a, b, or c
# [^abc]	Not a, b, or c
# [a-z]	Any lowercase letter
# {n,m}	Between n and m repetitions	{6,} = 6 or more
# *	0 or more
# +	1 or more
# ?	0 or 1
# ()	Grouping
```

```python
t1=re.findall(r'\b[a-zA-Z]{6,}\b',text)
print(t1)
print()

t2=re.findall(r'\b\d+\d*\b',text)
#\b\d+\.?\d*\b
print(t2)

t3=re.findall(r'\b[A-Z][a-zA-Z]*\b',text)
print(t3)
print()

t4=re.findall(r'\b[a-zA-Z]\b',text)
print(t4)
print()

t5=re.findall(r'\b[aeiouAEIOU][a-zA-Z]*\b',text)
print(t5)
print()
```

3. Use Regex SubsƟtuƟons (re.sub) to:
a. Replace email addresses with '<EMAIL>' placeholder.
b. Replace URLs with '<URL>' placeholder.
c. Replace phone numbers (formats: 123-456-7890 or +91 9876543210) with
'<PHONE>' placeholder.

```python
text = """Contact me at example.email@gmail.com or visit https://my-website.com.
Call me at 123-456-7890 or +91 9876543210. This isn't just any text—it’s a state-of-the-art NLP system.
Also, Pi is approximately 3.14."""

def cust(text):

    return re.findall( r"\b\w+(?:[-']\w+)*\b|\d+\.\d+|\d+",text)

print(cust(text))
```

```python
print(re.sub(r'\S+@\S+','<EMAIL>',text))
print()
print(re.sub(r'(\+91\s?)?\d{10}','<Phone>',text))
print()
print(re.sub(r'https?://[^\s]+', '<URL>', text))
```
